```python
def Calculate_Sales(Dataframe):
    Dataframe["Gross_Amount"]= (Dataframe["Quantity"]*Dataframe["Unit_Price"])
    Dataframe["Discount_Amount"]=(Dataframe["Gross_Amount"]*Dataframe["Discount"])
    Dataframe["Final_Amount"]= (Dataframe["Gross_Amount"] -Dataframe["Discount_Amount"])
    return Dataframe
```

```python
try:
  df=pd.read_csv("/content/raw_sales.csv")
  print("File loaded Sucessfully")
except FileNotFoundError:
  print("csv file not found")
except Exception as e:
  print("unexpected error,e")
```

```python
print("rows",df.shape[0])
print("columns",df.shape[1])
print("/nDuplicate records:")
print(df.duplicated().sum())
print("missing values")
print(df.isnull().sum())
```

```python
df.to_csv("Cleaned_Sales_data.csv",index=False)
```

```python
df = Calculate_Sales(df)
Category_Sales=(df.groupby("Category")["Final_Amount"].sum().reset_index())
Category_Sales.to_csv("Category_Sales_Summary.csv",index=False)
```

```python
Customer_Sales=(df.groupby("Customer")["Final_Amount"].sum().reset_index())
Customer_Sales.to_csv("Customer_Sales_Summary.csv",index=False)
```

```python
import matplotlib.pyplot as plt
Customer_Sales=(df.groupby("Customer")["Final_Amount"].sum().sort_values(ascending=False).head(10).reset_index())
Customer_Sales.columns=["Customer","total_sales"]
plt.figure(figsize=(10,6))
plt.barh(Customer_Sales["Customer"],Customer_Sales["total_sales"])
plt.xlabel("Customer")
plt.ylabel("Total Sales")
plt.title("Top 10 Customers by Sales")
plt.tight_layout()
plt.show()
```

```python
import matplotlib.pyplot as plt
Category_Sales=(df.groupby("Category")["Final_Amount"].sum().sort_values(ascending=False).head(10).reset_index())
Category_Sales.columns=["Category","total_sales"]
plt.figure(figsize=(8,5))
plt.bar(Category_Sales["Category"],Category_Sales["total_sales"])
plt.xlabel("Category")
plt.ylabel("Total Sales")
plt.title("Top 10 Category by Sales")
plt.tight_layout()
plt.show()
```

```python
plt.figure(figsize=(7,7))
plt.pie(Category_Sales["total_sales"],labels = Category_Sales["Category"],autopct="%1.1f%%")
plt.title("Category Sales Distribution")
plt.show()
```

```python
from matplotlib import figure
Daily_sales=(df.groupby("Order_Date")["Final_Amount"].sum().reset_index().sort_values("Order_Date"))
plt.figure(figsize=(12,6))
plt.plot(Daily_sales["Order_Date"],Daily_sales["Final_Amount"],marker="*",linestyle="-")
plt.xlabel("Order Date")
plt.ylabel("Total Sales")
plt.title("Daily Sales Trend")
```
